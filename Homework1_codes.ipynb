```python
#manually defining relevance set to determine precision
import os
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download necessary NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')

# Preprocess the text (tokenization, stopword removal)
def preprocess(text):
    stop_words = set(stopwords.words('english'))
    tokens = word_tokenize(text.lower())  # Tokenize and lowercase
    filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]  # Remove stopwords and non-alphanumeric words
    return filtered_tokens

# Load all .txt files into a dictionary
def load_documents(directory):
    documents = {}
    for filename in os.listdir(directory):
        if filename.endswith(".txt"):
            with open(os.path.join(directory, filename), 'r', encoding='utf-8') as file:
                documents[filename] = preprocess(file.read())
    return documents

# Build the Boolean term-document matrix
def build_boolean_model(documents):
    all_terms = set()
    for doc in documents.values():
        all_terms.update(doc)

    term_document_matrix = {}
    for term in all_terms:
        term_document_matrix[term] = {}
        for doc_name, doc_terms in documents.items():
            term_document_matrix[term][doc_name] = 1 if term in doc_terms else 0
    return term_document_matrix

# Boolean retrieval: Find retrieved documents (AND logic)
def boolean_retrieval(term_document_matrix, documents, query):
    query_terms = preprocess(query)

    if not query_terms:
        return set()  # No valid query terms

    # Start with all documents and refine with AND logic
    retrieved_docs = set(documents.keys())

    for term in query_terms:
        if term in term_document_matrix:
            term_docs = {doc for doc, present in term_document_matrix[term].items() if present == 1}
            retrieved_docs &= term_docs  # AND operation: only keep docs that have this term
        else:
            return set()  # If any term is missing, return an empty set (no docs match all terms)

    return retrieved_docs

# Manually define relevant documents (Ground Truth) — Adjust this after checking output
def manually_define_relevant_docs():
    return {"artc17.txt", "artc74.txt", "artc67.txt","artc44.txt","artc43.txt","artc41.txt","artc46.txt","artc42.txt","artc69.txt"}  # Change this manually based on retrieved results ,"artc73.txt","artc84.txt","artc23.txt","artc20.txt","artc22.txt","artc23.txt","artc92.txt"

# Calculate Precision & Recall
def calculate_precision_recall(retrieved, relevant):
    retrieved = set(retrieved)
    relevant = set(relevant)

    true_positives = len(retrieved & relevant)  # Correctly retrieved
    precision = true_positives / len(retrieved) if retrieved else 0.0
    recall = true_positives / len(relevant) if relevant else 0.0

    return precision, recall

# Load documents
directory = 'news_articles'  # Change this if needed
documents = load_documents(directory)
boolean_model = build_boolean_model(documents)

# Take the query as input
user_query = input("Enter your search query: ").strip()

# Retrieve documents using Boolean retrieval (AND logic)
retrieved_docs = boolean_retrieval(boolean_model, documents, user_query)

# Manually adjust relevant documents after checking output
relevant_docs = manually_define_relevant_docs()

# Compute Precision & Recall
precision, recall = calculate_precision_recall(retrieved_docs, relevant_docs)

# Output results
print(f"\nQuery: {user_query}")
print(f"Precision: {precision:.2f}, Recall: {recall:.2f}")
print("\n Retrieved Documents:")
print(retrieved_docs)

print("\n Relevant Documents (Ground Truth) - Adjusted Manually:")
print(relevant_docs)
```
